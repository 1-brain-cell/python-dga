## Functions

ตัวอย่าง function เล็ก ๆ น้อย ๆ

````python
def calculate_bmi(weight, height_cm):
    """คำนวณ BMI และคืนค่าสถานะ"""
    height_m = height_cm / 100
    bmi = weight / (height_m ** 2)

    if bmi < 18.5:
        status = "Underweight"
    elif bmi < 25:
        status = "Normal"
    elif bmi < 30:
        status = "Overweight"
    else:
        status = "Obese"

    return round(bmi, 2), status
````

````python
bmi_val, bmi_status = calculate_bmi(70, 175)
print(f"BMI: {bmi_val}\nStatus: {bmi_status}")
````

````python
def power(base, exponent=2):
    """ยกกำลัง — exponent default คือ 2"""
    return base ** exponent

print(power(3))     # 3^2 = 9
print(power(3, 3))  # 3^3 = 27
print(power(2, 10)) # 2^10 = 1024
````

## Class

**Class** คือแม่พิมพ์ หรือ blueprint สำหรับสร้างสิ่งของ  
**Object** คือสิ่งของที่สร้างขึ้นมาจาก Class นั้น

ตัวอย่าง เช่น Class Student สำหรับเก็บข้อมูลและวิเคราะห์นักเรียน

````python
class Student:
    def __init__(self, student_id, name, scores):
        self.student_id = student_id
        self.name       = name
        self.scores     = scores  # list ของคะแนน

    def average(self):
        return sum(self.scores) / len(self.scores)

    def grade(self):
        avg = self.average()
        if avg >= 80: return "A"
        elif avg >= 70: return "B"
        elif avg >= 60: return "C"
        elif avg >= 50: return "D"
        else: return "F"

    def report(self):
        print(f"[{self.student_id}] {self.name}")
        print(f"  คะแนน: {self.scores}")
        print(f"  เฉลี่ย: {self.average():.1f} \n  เกรด: {self.grade()}")
````

````python
# สร้างนักเรียนหลายคน
students = [
    Student("ST001", "Alice",  [85, 90, 78, 92]),
    Student("ST002", "Bob",    [70, 65, 72, 68]),
    Student("ST003", "Carol",  [95, 88, 91, 97]),
]
````

````python
for s in students:
    s.report()
    print()

# หาคนที่คะแนนสูงที่สุด
best_student = students[0]          # สมมติคนแรกดีที่สุดก่อน
for s in students:
    if s.average() > best_student.average():
        best_student = s            # ถ้าเจอคนที่ดีกว่า ให้อัปเดต

print(f"คะแนนสูงสุด: {best_student.name} ({best_student.average():.1f})")
````
