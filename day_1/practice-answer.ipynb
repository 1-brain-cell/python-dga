# **คิดเงินหน่อยค่ะ**
**หัวข้อ:** Conditional (if-elif-else) / Arithmetic

รับอายุลูกค้าและจำนวนจาน คำนวณราคาตามเงื่อนไข:
- อายุ < 5 ปี : ฟรี
- อายุ 5–11 ปี : จานละ 50 บาท
- อายุ 12–59 ปี : จานละ 100 บาท
- อายุ ≥ 60 ปี : จานแรกฟรี จานถัดไปจานละ 50 บาท

**Testcases:**

Input: อายุ `62`, จำนวนจาน `3` --> Output: `Pay 100 baht`

Input: อายุ `8`, จำนวนจาน `2` --> Output: `Pay 100 baht`

Input: อายุ `3`, จำนวนจาน `5` --> Output: `Pay 0 baht`


````python
age   = int(input())
plate = int(input())

if age < 5:
    total = 0
elif age < 12:
    total = plate * 50
elif age < 60:
    total = plate * 100
else:
    if plate <= 1:
        total = 0
    else:
        total = (plate - 1) * 50

print("Pay %d baht" % total)
````

# **เกมนับเลข (Fizz Buzz)**

จงเขียนโปรแกรม Python ที่รับจำนวนเต็มบวกจากผู้ใช้ใส่ตัวแปร num โปรแกรมควรแสดงตัวแปร output ที่มีค่าเป็นตัวเลขตั้งแต่ 1 ถึง num ยกเว้นในกรณีต่อไปนี้:
*   ถ้า num หารด้วย 3 ลงตัว ให้แทนที่ output ด้วยสตริง "Fizz"
*   ถ้า num หารด้วย 5 ลงตัว ให้แทนที่ output ด้วยสตริง "Buzz"
*   ถ้า num หารด้วยทั้ง 3 และ 5 ลงตัว ให้แทนที่ output ด้วยสตริง "Fizz Buzz"

<br /><br />
**Testcases:**

Input (ใส่จำนวนเต็มบวก ตัวแปร "num"): 15

Expected Output (แสดงตัวแปร "output"): <br />
1<br />
2<br />
Fizz<br />
4<br />
Buzz<br />
Fizz<br />
7<br />
8<br />
Fizz<br />
Buzz<br />
11<br />
Fizz<br />
13<br />
14<br />
Fizz Buzz

````python
n = int(input("ใส่จำนวนเต็มบวก: "))

for i in range(1, n + 1):
    if i % 3 == 0 and i % 5 == 0:
        output = "Fizz Buzz"
    elif i % 3 == 0:
        output = "Fizz"
    elif i % 5 == 0:
        output = "Buzz"
    else:
        output = i
    print(output)
````

# **ฟังก์ชัน BMI**
**หัวข้อ:** Function / Float / Conditional

เขียนฟังก์ชัน `calculate_bmi(weight, height_cm)` รับน้ำหนัก (kg) และส่วนสูง (cm) แล้วคืนค่า BMI และระดับ:
- BMI < 18.5 → `Underweight`
- 18.5–24.9 → `Normal`
- 25.0–29.9 → `Overweight`
- ≥ 30 → `Obese`

**Testcases:**

Input: น้ำหนัก `70`, ส่วนสูง `175` --> Output: `BMI: 22.86 | Normal`

Input: น้ำหนัก `90`, ส่วนสูง `165` --> Output: `BMI: 33.06 | Obese`


````python
def calculate_bmi(weight, height_cm):
    height_m = height_cm / 100
    bmi = weight / (height_m ** 2)

    if bmi < 18.5:
        status = "Underweight"
    elif bmi < 25.0:
        status = "Normal"
    elif bmi < 30.0:
        status = "Overweight"
    else:
        status = "Obese"

    return round(bmi, 2), status

weight    = float(input())
height_cm = float(input())

bmi_val, bmi_status = calculate_bmi(weight, height_cm)
print("BMI: %s | %s" % (bmi_val, bmi_status))
````

# **มองภาพรวม**
**หัวข้อ:** List / Sorting / Slicing / Built-in Functions

รับคะแนนนักเรียนหลายคน คั่นด้วย comma (เช่น `85,90,78,92,65`) แล้วแสดง:
- คะแนนสูงสุดและต่ำสุด
- ค่าเฉลี่ย (ทศนิยม 2 ตำแหน่ง)
- 3 อันดับแรก (เรียงจากมากไปน้อย)

**Testcase:**

Input: `85,90,78,92,65`

Expected Output:
```
Max: 92, Min: 65
Average: 82.00
Top 3: [92, 90, 85]
```


````python
scores = list(map(int, input().split(",")))

max_s = max(scores)
min_s = min(scores)
avg   = sum(scores) / len(scores)
top3  = sorted(scores, reverse=True)[:3]

print("Max: %d, Min: %d" % (max_s, min_s))
print("Average: %.2f"    % avg)
print("Top 3: "          + str(top3))
````

# **การเข้ารหัส (Encoding)**

จงเขียนฟังก์ชัน Python ชื่อ encode(string) ที่รับสตริง string เป็นอินพุต และคืนค่าเป็นสตริงใหม่ที่แต่ละตัวอักษรถูกแทนที่ด้วยความถี่ของตัวอักษรนั้นในสตริงเดิม ตามด้วยตัวอักษรนั้นเอง

ตัวอย่างเช่น สตริงอินพุต "aaabbc" ควรคืนค่าเป็นสตริง "3a2b1c"

<br /><br />
**Testcases:**

Input #1 (ใส่สตริง): aaabbc

Expected Output #1 (แสดงสตริงที่เข้ารหัสแล้ว): 3a2b1c

<br />
Input #2 (ใส่สตริง): abcabcabc

Expected Output #2 (แสดงสตริงที่เข้ารหัสแล้ว): 1a1b1c1a1b1c1a1b1c

<br />
Input #3 (ใส่สตริง): AaaBbcccc

Expected Output #3 (แสดงสตริงที่เข้ารหัสแล้ว): 1A2a1B1b4c

````python
def encode(string):
    if not string:
        return ""

    result = ""
    count = 1

    for i in range(1, len(string)):
        if string[i] == string[i - 1]:
            count += 1
        else:
            result += str(count) + string[i - 1]
            count = 1

    result += str(count) + string[-1]
    return result

# ทดสอบ
text = input("ใส่สตริง: ")
print(encode(text))
````

# **นับความถี่ของคำ**
**หัวข้อ:** Dictionary / Loop / String / Data Analysis

เขียนฟังก์ชัน `word_freq(text)` รับประโยคภาษาอังกฤษ นับจำนวนครั้งที่แต่ละคำปรากฏ (ไม่สนตัวพิมพ์เล็ก-ใหญ่) แล้วแสดง **3 อันดับแรก** เรียงจากมากไปน้อย

**Testcases:**

Input: `the cat sat on the mat the cat`

Expected Output:
```
the: 3
cat: 2
sat: 1
```

Input: `Data is the new oil data drives decisions`

Expected Output:
```
data: 2
is: 1
the: 1
```


````python
def word_freq(text):
    words = text.lower().split()

    freq = {}
    for word in words:
        freq[word] = freq.get(word, 0) + 1

    top3 = sorted(freq.items(), key=lambda x: x[1], reverse=True)[:3]

    for word, count in top3:
        print("%s: %d" % (word, count))

text = input()
word_freq(text)
````

````python
# หรือใช้วิธีนี้
def word_freq(text):
    words = text.lower().split()

    freq = {}
    for word in words:
        freq[word] = freq.get(word, 0) + 1

    result = []
    freq_copy = dict(freq)

    for _ in range(3):
        best_word  = ""
        best_count = -1
        for word, count in freq_copy.items():
            if count > best_count:
                best_word  = word
                best_count = count
        result.append((best_word, best_count))
        del freq_copy[best_word]

    for word, count in result:
        print("%s: %d" % (word, count))

text = input()
word_freq(text)
````

# **บัญชีธนาคาร (Bank Account)**

สร้างคลาส BankAccount ที่มีเมธอดต่อไปนี้:
*   เมธอด **\_\_init\_\_** ที่รับยอดเงินเริ่มต้นเป็นอาร์กิวเมนต์และกำหนดยอดเงินของบัญชี
*   เมธอด **deposit** ที่รับจำนวนเงินเป็นอาร์กิวเมนต์และเพิ่มเข้าไปในยอดเงินของบัญชี
*   เมธอด **withdraw** ที่รับจำนวนเงินเป็นอาร์กิวเมนต์และหักออกจากยอดเงินของบัญชี แต่เฉพาะในกรณีที่บัญชีมีเงินเพียงพอเท่านั้น
*   เมธอด **get_balance** ที่ส่งคืนยอดเงินปัจจุบันของบัญชี

<br /><br />
**Testcases:**

Input #1 (ใส่ยอดเงินเริ่มต้น จำนวนเงินที่ฝาก และจำนวนเงินที่ถอน): <br />
2500<br />
500<br />
1000<br />

Expected Output #1 (แสดงยอดเงินคงเหลือในบัญชีปัจจุบัน): 2000.00

<br />
Input #2 (ใส่ยอดเงินเริ่มต้น จำนวนเงินที่ฝาก และจำนวนเงินที่ถอน): <br />
999<br />
1000<br />
2000<br />

Expected Output #2 (แสดงยอดเงินคงเหลือในบัญชีปัจจุบัน): Insufficient funds

````python
class BankAccount:
    def __init__(self, balance):
        self.balance = balance

    def deposit(self, amount):
        self.balance += amount

    def withdraw(self, amount):
        if amount > self.balance:
            return False   # ถอนไม่สำเร็จ
        else:
            self.balance -= amount
            return True    # ถอนสำเร็จ

    def get_balance(self):
        return self.balance
````

````python
initial  = float(input("ยอดเงินเริ่มต้น: "))
deposit  = float(input("จำนวนเงินที่ฝาก: "))
withdraw = float(input("จำนวนเงินที่ถอน: "))

acc = BankAccount(initial)
acc.deposit(deposit)
success = acc.withdraw(withdraw)

if success:
    print(f"{acc.get_balance():.2f}")
else:
    print("Insufficient funds")
````
